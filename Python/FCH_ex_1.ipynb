```python
# import libraries
import ssav_helpers
import ssav
from ssav_setup import *
from ssav_helpers import *
from ssav import *
import numpy as np
import matplotlib.pyplot as plt
from numpy.fft import fftn, ifftn, fftfreq
from dataclasses import dataclass
```

```python
L = np.array([2*np.pi, 2*np.pi])
N = np.array([128, 128])

Grid = generate_Grid(L, N, dim=2)

Time = TimeDomain(dt_min=1e-5, 
                  dt_max=0.01, 
                  t0=0, 
                  tf=5)

Para = Parameters(model="FCH",
                  eps=0.18, 
                  m=0.0, 
                  S=1.0, 
                  B=L[0]**5)

u = (2 * np.exp(np.sin(Grid.xx - np.pi) + np.sin(Grid.yy - np.pi) - 2)
    + 2.2 * np.exp(-np.sin(Grid.xx - np.pi) - np.sin(Grid.yy - np.pi) - 2) - 1)
```

```python
img = plt.imshow(u, cmap = 'bwr')
img.set_clim(-1,1)

plt.colorbar
plt.show()
```

```python
Results = ssav(Grid, Time, Para, u, dim=2)
```

```python
img = plt.imshow(Results.u, cmap = 'bwr')
img.set_clim(-1,1)

plt.colorbar
plt.show()

f1 = plt.figure()
plt.plot(Results.tt, Results.Eu)
f1.show

f2 = plt.figure()
plt.semilogy(np.array(Results.tt[1:]), Results.dt)
f2.show
```
